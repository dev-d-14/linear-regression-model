```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('Salary_Data.csv')
data.head()
```

```python
x_train=data['YearsExperience'].values
y_train=data['Salary'].values
plt.scatter(x_train,y_train)
plt.title("scatter plot")
plt.xlabel("experience")
plt.ylabel("salary")
plt.Figure(figsize=(6,4))
plt.show()
```

```python
def cost_function(x,y,m,c):
    cost_sum=0
    for i in range(len(x)):
        f = (m*x[i] + c)
        cost=(f-y[i])**2
        cost_sum += cost
    total_cost=(1/(2*m)) * cost_sum
    return total_cost
```

```python
def gradient_function(x,y,m,c):
    dc_m=0
    dc_c=0

    for i in range(len(x)):
        f = (m*x[i] + c)
        dc_m += (f - y[i]) * x[i]
        dc_c += (f - y[i]) 
    dc_m /= len(x)
    dc_c /= len(x)

    return dc_m,dc_c
```

```python
def gradient_descent(x,y,alpha,iterations):
    m=0
    c=0

    for i in range(iterations):
        dc_m,dc_c = gradient_function(x,y,m,c)

        m = m - alpha * dc_m
        c = c - alpha * dc_c
    return m,c
```

```python
L=0.01
epochs=1000

final_m,final_c = gradient_descent(x_train,y_train,L,epochs)

print("m: "+str(final_m)+" c: "+str(final_c))
```
